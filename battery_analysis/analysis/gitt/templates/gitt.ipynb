# Diffusion coefficients (GITT and √t pulse method)

Set in **`gitt.yaml`** (same pulse folders as `tafel.yaml`).

* **Classic GITT** (Weppner–Huggins) from each titration step and its relaxation. It is only valid while τD/L² ≪ 1 — see `gitt_validity_tauD_L2`; invalid states are drawn hollow.
* **√t method** (the workbook method) from one short pulse per state: D = 4/π (I V_M/(F S))² ((dE/dx)/(dE/d√t))². `dEdx: excel` reproduces the workbook's difference scheme.

Outputs: `Exports/GITT_<batch>.xlsx` · `Figures/GITT/<batch>/` · data in `Exports/PlotData/GITT/<batch>/`.

```python
%matplotlib inline
import matplotlib.pyplot as plt
import pandas as pd
from battery_analysis.project import Project

project = Project("..")
project.use("gitt", "gitt.yaml")
BATCH = "Batch1"                     # the batch this notebook works on
print("batches:", list(project._get("gitt").batches))
```

## 1. Analyse (writes the Excel file)

```python
samples = project.run_gitt(batch=BATCH)
table = pd.concat([s.table for s in samples], ignore_index=True)
cols = ["sample", "state", "x", "OCV_V", "tau_s", "dEs_V", "dEt_V", "D_gitt_cm2_s", "gitt_validity_tauD_L2",
        "sqrt_t_I_A", "dE_dsqrt_t", "sqrt_t_r2", "dE_dx", "D_sqrt_t_cm2_s"]
table[[c for c in cols if c in table]]
```
