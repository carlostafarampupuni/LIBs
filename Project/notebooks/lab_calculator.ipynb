# Lab Calculator — electrode / slurry / N-P ratio

This is the calculator side of the toolkit, not a raw-cycler-data
technique: no import from `Data/`, no database, just the numbers
you measure at the bench (foil/disc weights, active-material
fraction, slurry recipe ratios) run through `battery_analysis`'s
`lab_electrode` / `lab_slurry` / `lab_np_ratio` calculators.

**Batch mode** (below): one YAML file lists every electrode,
slurry, and N/P pair for a formulation session; `run_lab_batch()`
computes all of them in one go and writes the results as CSVs into
this project's `exports/` folder — a permanent, diffable record
instead of another spreadsheet tab.

**One-off mode** (further down): compute a single electrode,
slurry, or N/P pair inline, without a YAML file — handy for a
quick check mid-experiment.

## Batch mode

Runs `lab_batch_example.yaml` (at the project root) by default —
copy that file, rename it (e.g. `lab_batch.yaml`), and fill in your
own electrodes/slurries/np_pairs before pointing this cell at your
real formulation. Every section (`electrodes:`, `slurries:`,
`np_pairs:`) is optional — a file with only `electrodes:` just
skips the slurry/N-P CSVs.

## One-off mode

No YAML file — compute a single electrode, slurry, or N/P pair
directly. Useful for a quick check without editing a batch file.

```python
from battery_analysis.lab.lab_electrode import ElectrodeMeasurement

electrode = ElectrodeMeasurement(
    name="Example (Cathode)",
    foil_weight_g=0.0085,
    foil_thickness_mm=0.02,
    disc_weight_g=0.0205,
    disc_thickness_mm=0.058,
    am_fraction=0.8,
    theoretical_capacity_mAhg=170,
    diameter_mm=14.0,
)
electrode.summary()
```

```python
from battery_analysis.lab.lab_slurry import SlurryComponent, SlurryFormulation

slurry = SlurryFormulation(
    total_dry_solids_g=3.0,
    extra_solvent_g=6.0,
    components=[
        SlurryComponent(name="LFP", ratio=0.8),
        SlurryComponent(name="Super-P", ratio=0.1),
        SlurryComponent(name="PVDF", ratio=0.1, concentration_fraction=0.10),  # 10% PVDF-in-NMP stock
    ],
)
slurry.summary()
```
