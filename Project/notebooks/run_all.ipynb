## Run everything
One click instead of one notebook per technique. `project.use(...)` picks which techniques' YAMLs to load (comment a line out to skip that technique entirely); `project.run_all()` then imports, analyzes/fits, and plots EVERY batch declared in each of those YAMLs — including any batch `auto_discover:` picked up automatically from a new folder under `Data/` (currently EIS, Cycle, Rate — see the `auto_discover:` block near the top of each of those YAMLs). One bad batch (missing files, wrong format) is caught and reported, not raised — it won't stop every other batch from running. See the `run_all()` docstring in `project.py` for the full picture, and the README's "Run everything at once" section.

```python
# show plots inline in this notebook / PyCharm's notebook window
%matplotlib inline
import sys
sys.path.insert(0, "..")
from battery_analysis.project import Project

project = Project("..")
project.use("eis", "eis.yaml")
project.use("drt", "drt.yaml")
project.use("cycle", "cycle.yaml")
project.use("rate", "rate.yaml")
project.use("gcd", "gcd.yaml")
project.use("xps", "xps.yaml")
project.use("dqdv", "dqdv.yaml")
project.use("ric", "ric.yaml")
project.use("sem", "sem.yaml")
project.use("ftir", "ftir.yaml")
project.use("cv", "cv.yaml")
```

### Preview — what would run
No database writes, no figures — just the technique/batch pairs `run_all()` is about to process. Good place to check a new `auto_discover:`-picked-up folder actually shows up before committing to a real run.

```python
plan = project.run_all(dry_run=True)
```

### The real run
This is the one that imports, analyzes, and plots everything above. Takes a few minutes with this many batches — that's expected. `stop_on_error=True` instead lets a specific batch's real traceback stop the run immediately, useful while debugging one batch rather than reading the summary at the end.

```python
results = project.run_all()
```
